# 🌊 Why Are Submarines Shaped Like Cigars?

## A computational experiment in fluid drag and underwater vehicle shape

**Research question:** How does streamlining affect the drag and propulsion power required by an underwater vehicle?

The goal is not to reproduce a real submarine's exact hydrodynamics. Instead, this project uses a simplified drag model to investigate why a streamlined body can be advantageous.

## 1. The physics

A simplified drag model is:

$$D = \\frac{1}{2}\\rho V^2 C_D A$$

where D is drag force, ρ is water density, V is speed, Cᴅ is drag coefficient, and A is reference frontal area.

The approximate power required to overcome drag is:

$$P = DV$$

Therefore, if the other variables remain approximately constant, drag grows with V² and idealized drag power grows with V³.

## 2. Conceptual experiment

I compare three simplified bodies: a sphere, a blunt cylinder, and a streamlined teardrop.

The reference frontal area is held constant for the basic comparison. The drag coefficients are **illustrative educational assumptions**, not CFD results or measurements of real submarines.

```python
import numpy as np
import matplotlib.pyplot as plt

rho=1025
A=12
speeds=np.linspace(1,20,100)
Cd={'Sphere':0.47,'Blunt cylinder':1.05,'Streamlined teardrop':0.12}
for name,cd in Cd.items():
    D=0.5*rho*speeds**2*cd*A
    plt.plot(speeds,D,label=name)
plt.xlabel('Speed (m/s)')
plt.ylabel('Estimated drag (N)')
plt.title('Conceptual drag comparison')
plt.legend()
plt.show()
```

## 3. Why speed changes everything

If Cᴅ, A and water density remain approximately constant:

$$D \\propto V^2$$

and:

$$P \\propto V^3$$

Therefore doubling speed gives approximately four times the drag force and eight times the idealized drag power.

```python
v1=5
v2=10
print('Drag ratio:',(v2/v1)**2)
print('Power ratio:',(v2/v1)**3)
```

## 4. What the model can—and cannot—tell us

The model explains the role of frontal area and drag coefficient, but it does not capture full submarine hydrodynamics.

A real design must consider Reynolds number, boundary layers, skin friction, flow separation, appendages, propulsor interactions, structural constraints, buoyancy, noise, maneuverability and other requirements.

This is therefore a **first-principles engineering experiment**, not a design specification.

## 5. Engineering conclusion

The streamlined shape is not about appearance. A streamlined underwater body can reduce pressure/form drag by delaying flow separation and maintaining smoother flow around the vehicle.

However, reducing drag is only one objective. A real submarine must balance hydrodynamic efficiency with volume, strength, stability, maneuverability, equipment and other requirements.
